## 1. Definicion de Inteligencia Artificial

> "Esfuerzo por automatizar las tareas intelectuales normalmente realizadas por humanos"

La Inteligencia Artificial es un campo multidisciplinario que busca crear sistemas capaces de realizar tareas que, hasta hace poco, requerian inteligencia humana.

### 1.1. Evolucion Historica de la IA

| **Año** | **Hito** |
|---------|----------|
| **1943** | Redes Neuronales (McCulloch-Pitts) |
| **1957-62** | Perceptron (Frank Rosenblatt) |
| **1970-86** | Retropropagacion (Backpropagation) |
| **1979-1998** | CNN, MNIST, LSTM, RNN bidireccional |
| **2006** | "Deep Learning" (Hinton) |
| **2009** | ImageNet + AlexNet |
| **2014** | GANs (Redes Generativas Adversarias) |
| **2016-17** | AlphaGo, AlphaZero |
| **2017-19** | Transformers (BERT, GPT) |
| **2020-26** | GPT3, GPT4(multimodal),Agentes |
### 1.2. Paradigmas de la IA

| **Paradigma** | **Descripcion** | **Ejemplos** |
|--------------|-----------------|--------------|
| **Simbolica** | Razonamiento basado en reglas y logica | Sistemas expertos, Prolog |
| **Subsimbolica** | Aprendizaje desde datos sin reglas explicitas | Redes Neuronales |
| **Estadistica** | Modelos probabilisticos y estadisticos | Machine Learning |

### 1.3. Por que Ahora?

**Tres factores clave:**
1. **Datos**: Camaras de alta resolucion, celulares, redes sociales, experimentos cientificos
2. **Software**: PyTorch, TensorFlow, herramientas para implementar modelos
3. **Hardware**: Procesamiento en la nube, GPUs, FPGA, sistemas embebidos

## 3. Tipos de Aprendizaje

### 3.1. Aprendizaje Supervisado (Supervised Learning)

**Caracteristicas:**
- Datos etiquetados
- Retroalimentacion: Error de prediccion
- Objetivo: Predecir salida

**Sub-tipos:**
| **Tipo** | **Salida** | **Ejemplo** |
|----------|------------|-------------|
| **Regresion** | Continua | Predecir precio de vivienda |
| **Clasificacion** | Discreta | Identificar digitos manuscritos |

**Algoritmos comunes:**
- Regresion Lineal
- Regresion Logistica
- SVM (Maquinas de Soporte Vectorial)
- Arboles de Decision
- Redes Neuronales

### 3.2. Aprendizaje No Supervisado (Unsupervised Learning)

**Caracteristicas:**
- Sin etiquetas
- No hay retroalimentacion
- Objetivo: Encontrar patrones

**Sub-tipos:**
| **Tipo** | **Objetivo** | **Ejemplo** |
|----------|-------------|-------------|
| **Clustering** | Agrupar datos similares | Segmentacion de clientes |
| **Reduccion de dimensionalidad** | Proyectar a espacio menor | Visualizacion de datos |
| **Deteccion de anomalias** | Encontrar puntos atipicos | Deteccion de fraudes |

**Algoritmos comunes:**
- K-Means
- PCA (Analisis de Componentes Principales)
- t-SNE
- Autoencoders

### 3.3. Aprendizaje por Refuerzo (Reinforcement Learning)

**Caracteristicas:**
- Interaccion dinamica con el entorno
- Retroalimentacion: Recompensa/Castigo
- Objetivo: Maximizar recompensa futura

**Elementos clave:**
- **Agente**: El que aprende
- **Entorno**: Con el que interactua
- **Estado (s)**: Situacion actual
- **Accion (a)**: Lo que puede hacer
- **Recompensa (r)**: Senal de retroalimentacion

**Algoritmos comunes:**
- Q-Learning
- Deep Q-Networks (DQN)
- Policy Gradient

## 7. Vision General de un Proyecto de IA

### 7.1. Pipeline de Machine Learning

```
Datos → Preprocesamiento → Feature Engineering → Modelado → Evaluacion → Despliegue
```

### 7.2. Pasos de un Proyecto de IA

1. **Entendimiento del problema**: Definir objetivos y metricas
2. **Obtencion de datos**: Recolectar datos relevantes
3. **Exploracion de datos (EDA)**: Analizar y visualizar datos
4. **Preprocesamiento**: Limpiar y transformar datos
5. **Feature Engineering**: Crear caracteristicas relevantes
6. **Modelado**: Entrenar modelos
7. **Evaluacion**: Validar rendimiento
8. **Despliegue**: Poner en produccion
9. **Monitoreo**: Seguimiento continuo
