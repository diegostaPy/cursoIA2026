```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML

# Configuración
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 200)
pd.set_option('display.width', None)
pd.set_option('display.float_format', '{:.2f}'.format)

print("="*80)
print("📊 ESTADÍSTICAS POR PERÍODO - APROBADOS SOBRE INSCRIPTOS")
print("="*80)

# =============================================================================
# 1. CARGAR DATOS
# =============================================================================
print("\n📂 Cargando datos concatenados...")

ruta_archivo = 'D:\\ArchivosHDD\\GitHub\\cursoIA2026\\Semana_03\\Ejercicio\\datos\\datos_concatenados.csv'

try:
    df = pd.read_csv(ruta_archivo)
    print(f"✅ Datos cargados: {len(df):,} registros")
except FileNotFoundError:
    print(f"❌ No se encontró el archivo: {ruta_archivo}")
    exit()

# =============================================================================
# 2. PREPROCESAMIENTO
# =============================================================================
print("\n🔄 Preprocesando datos...")

# Estado de aprobación
if 'Aprobado' in df.columns:
    df['Aprobado_Num'] = df['Aprobado'].map({'S': 1, 'N': 0})

print(f"✅ Preprocesamiento completado")

# =============================================================================
# 3. ESTADÍSTICAS POR PERÍODO (AÑO - SEMESTRE)
# =============================================================================
print("\n" + "="*80)
print("📊 ESTADÍSTICAS POR PERÍODO")
print("="*80)

# Agrupar por año y semestre
estadisticas_periodo = df.groupby(['Anho', 'Semestre']).agg({
    'ALUMNO_ID': 'count',
    'Aprobado_Num': 'sum' if 'Aprobado_Num' in df.columns else 'count'
}).reset_index()

estadisticas_periodo.columns = ['Año', 'Semestre', 'Inscriptos', 'Aprobados']

if 'Aprobado_Num' in df.columns:
    estadisticas_periodo['Reprobados'] = estadisticas_periodo['Inscriptos'] - estadisticas_periodo['Aprobados']
    estadisticas_periodo['%_Aprobacion'] = (estadisticas_periodo['Aprobados'] / 
                                             estadisticas_periodo['Inscriptos'] * 100).round(2)
    estadisticas_periodo['%_Reprobacion'] = 100 - estadisticas_periodo['%_Aprobacion']

# Ordenar
estadisticas_periodo = estadisticas_periodo.sort_values(['Año', 'Semestre'])

# Crear período como string
estadisticas_periodo['Período'] = estadisticas_periodo['Año'].astype(str) + '-S' + estadisticas_periodo['Semestre'].astype(str)

print("\n📋 ESTADÍSTICAS POR PERÍODO:")
print("-" * 80)
display(estadisticas_periodo[['Período', 'Inscriptos', 'Aprobados', 'Reprobados', '%_Aprobacion', '%_Reprobacion']])

# =============================================================================
# 4. MATERIAS POR SEMESTRE
# =============================================================================
print("\n" + "="*80)
print("📚 LISTADO DE MATERIAS POR SEMESTRE")
print("="*80)

# Agrupar por año, semestre y materia
materias_por_semestre = df.groupby(['Anho', 'Semestre', 'Asignatura', 'Cod.Asign']).agg({
    'ALUMNO_ID': 'count',
    'Aprobado_Num': 'sum' if 'Aprobado_Num' in df.columns else 'count'
}).reset_index()

materias_por_semestre.columns = ['Año', 'Semestre', 'Materia', 'Código', 'Inscriptos', 'Aprobados']

if 'Aprobado_Num' in df.columns:
    materias_por_semestre['%_Aprobacion'] = (materias_por_semestre['Aprobados'] / 
                                              materias_por_semestre['Inscriptos'] * 100).round(2)

# Ordenar
materias_por_semestre = materias_por_semestre.sort_values(['Año', 'Semestre', 'Materia'])
materias_por_semestre['Período'] = materias_por_semestre['Año'].astype(str) + '-S' + materias_por_semestre['Semestre'].astype(str)

# Mostrar por período
periodos = materias_por_semestre['Período'].unique()

for periodo in sorted(periodos):
    print(f"\n{'='*80}")
    print(f"📚 {periodo}")
    print('='*80)
    
    df_periodo = materias_por_semestre[materias_por_semestre['Período'] == periodo]
    
    # Mostrar solo materias con inscriptos > 0
    df_periodo = df_periodo[df_periodo['Inscriptos'] > 0]
    
    # Crear tabla resumen
    tabla_materias = df_periodo[['Materia', 'Código', 'Inscriptos', 'Aprobados', '%_Aprobacion']].copy()
    tabla_materias = tabla_materias.sort_values('%_Aprobacion', ascending=False)
    
    # Mostrar estadísticas del período
    total_inscriptos = df_periodo['Inscriptos'].sum()
    total_aprobados = df_periodo['Aprobados'].sum()
    total_porcentaje = (total_aprobados / total_inscriptos * 100) if total_inscriptos > 0 else 0
    
    print(f"\n📊 Resumen del período:")
    print(f"   • Total inscriptos: {total_inscriptos:,}")
    print(f"   • Total aprobados: {total_aprobados:,}")
    print(f"   • Tasa de aprobación: {total_porcentaje:.2f}%")
    print(f"   • Total materias: {len(df_periodo)}")
    
    print(f"\n📋 Listado de materias:")
    display(tabla_materias)

# =============================================================================
# 5. RESUMEN POR AÑO
# =============================================================================
print("\n" + "="*80)
print("📊 RESUMEN POR AÑO")
print("="*80)

resumen_anho = df.groupby('Anho').agg({
    'ALUMNO_ID': 'count',
    'Aprobado_Num': 'sum' if 'Aprobado_Num' in df.columns else 'count'
}).reset_index()

resumen_anho.columns = ['Año', 'Inscriptos', 'Aprobados']

if 'Aprobado_Num' in df.columns:
    resumen_anho['Reprobados'] = resumen_anho['Inscriptos'] - resumen_anho['Aprobados']
    resumen_anho['%_Aprobacion'] = (resumen_anho['Aprobados'] / 
                                     resumen_anho['Inscriptos'] * 100).round(2)

print("\n📋 Resumen por año:")
display(resumen_anho)

# =============================================================================
# 6. RESUMEN POR SEMESTRE (TODOS LOS AÑOS)
# =============================================================================
print("\n" + "="*80)
print("📊 RESUMEN POR SEMESTRE (TODOS LOS AÑOS)")
print("="*80)

resumen_semestre = df.groupby('Semestre').agg({
    'ALUMNO_ID': 'count',
    'Aprobado_Num': 'sum' if 'Aprobado_Num' in df.columns else 'count'
}).reset_index()

resumen_semestre.columns = ['Semestre', 'Inscriptos', 'Aprobados']

if 'Aprobado_Num' in df.columns:
    resumen_semestre['Reprobados'] = resumen_semestre['Inscriptos'] - resumen_semestre['Aprobados']
    resumen_semestre['%_Aprobacion'] = (resumen_semestre['Aprobados'] / 
                                         resumen_semestre['Inscriptos'] * 100).round(2)

print("\n📋 Resumen por semestre (todos los años):")
display(resumen_semestre)

# =============================================================================
# 7. VISUALIZACIONES
# =============================================================================
print("\n" + "="*80)
print("📈 VISUALIZACIONES")
print("="*80)

# 7.1. Barras por período
print("\n📊 Registros y aprobados por período:")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Inscriptos vs Aprobados
axes[0].bar(estadisticas_periodo['Período'], estadisticas_periodo['Inscriptos'], 
            color='skyblue', alpha=0.7, label='Inscriptos')
axes[0].bar(estadisticas_periodo['Período'], estadisticas_periodo['Aprobados'], 
            color='green', alpha=0.7, label='Aprobados')
axes[0].set_xlabel('Período')
axes[0].set_ylabel('Número de Estudiantes')
axes[0].set_title('Inscriptos vs Aprobados por Período')
axes[0].legend()
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(True, alpha=0.3)

# Gráfico 2: Porcentaje de aprobación
axes[1].bar(estadisticas_periodo['Período'], estadisticas_periodo['%_Aprobacion'], 
            color='green', alpha=0.7)
axes[1].axhline(y=50, color='red', linestyle='--', alpha=0.5, label='Umbral 50%')
axes[1].set_xlabel('Período')
axes[1].set_ylabel('Porcentaje de Aprobación (%)')
axes[1].set_title('Porcentaje de Aprobación por Período')
axes[1].legend()
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(True, alpha=0.3)
axes[1].set_ylim(0, 100)

# Agregar valores en las barras
for i, v in enumerate(estadisticas_periodo['%_Aprobacion']):
    axes[1].text(i, v + 2, f'{v:.1f}%', ha='center', fontsize=10)

plt.tight_layout()
plt.show()

# 7.2. Top 10 materias con menor aprobación
print("\n📊 Top 10 materias con menor porcentaje de aprobación (todos los períodos):")
top_peores = materias_por_semestre.nsmallest(10, '%_Aprobacion')
top_peores = top_peores[top_peores['Inscriptos'] >= 10]  # Solo materias con al menos 10 estudiantes

plt.figure(figsize=(12, 6))
bars = plt.barh(top_peores['Materia'] + ' (' + top_peores['Período'] + ')', 
                top_peores['%_Aprobacion'], 
                color='red', alpha=0.7)
plt.xlabel('Porcentaje de Aprobación (%)')
plt.title('Top 10 Materias con Menor Porcentaje de Aprobación')
plt.xlim(0, 100)

# Agregar valores
for i, (bar, value) in enumerate(zip(bars, top_peores['%_Aprobacion'])):
    plt.text(value + 1, bar.get_y() + bar.get_height()/2, 
             f'{value:.1f}%', va='center')

plt.tight_layout()
plt.show()

# 7.3. Heatmap de aprobación por período
print("\n📊 Heatmap de aprobación por año y semestre:")
plt.figure(figsize=(10, 6))

# Preparar datos para heatmap
heatmap_data = df.pivot_table(index='Anho', columns='Semestre', 
                              values='Aprobado_Num', aggfunc='mean', fill_value=0) * 100

sns.heatmap(heatmap_data, annot=True, fmt='.1f', cmap='RdYlGn', 
            center=50, cbar_kws={'label': 'Porcentaje de Aprobación (%)'})
plt.title('Heatmap de Porcentaje de Aprobación por Año y Semestre')
plt.xlabel('Semestre')
plt.ylabel('Año')
plt.tight_layout()
plt.show()

# =============================================================================
# 8. EXPORTAR RESULTADOS
# =============================================================================
print("\n" + "="*80)
print("💾 EXPORTANDO RESULTADOS")
print("="*80)

estadisticas_periodo.to_csv('estadisticas_por_periodo.csv', index=False)
materias_por_semestre.to_csv('materias_por_semestre.csv', index=False)
resumen_anho.to_csv('resumen_por_anho.csv', index=False)
resumen_semestre.to_csv('resumen_por_semestre.csv', index=False)

print("✅ Archivos guardados:")
print("   • estadisticas_por_periodo.csv")
print("   • materias_por_semestre.csv")
print("   • resumen_por_anho.csv")
print("   • resumen_por_semestre.csv")

# =============================================================================
# 9. RESUMEN FINAL
# =============================================================================
print("\n" + "="*80)
print("📋 RESUMEN FINAL")
print("="*80)

if 'Aprobado_Num' in df.columns:
    total_inscriptos = len(df)
    total_aprobados = df['Aprobado_Num'].sum()
    porcentaje_general = (total_aprobados / total_inscriptos * 100)
    
    print(f"""
✅ ANÁLISIS COMPLETADO

📊 Estadísticas Generales:
   • Total de inscriptos: {total_inscriptos:,}
   • Total de aprobados: {total_aprobados:,}
   • Tasa de aprobación general: {porcentaje_general:.2f}%
   • Total de estudiantes únicos: {df['ALUMNO_ID'].nunique():,}
   • Total de materias: {df['Asignatura'].nunique():,}
   • Períodos: {df['Anho'].min()} - {df['Anho'].max()}
""")

# Mejor y peor período
if len(estadisticas_periodo) > 0:
    mejor_periodo = estadisticas_periodo.loc[estadisticas_periodo['%_Aprobacion'].idxmax()]
    peor_periodo = estadisticas_periodo.loc[estadisticas_periodo['%_Aprobacion'].idxmin()]
    
    print(f"""
📈 Mejor y Peor Período:
   • Mejor: {mejor_periodo['Período']} - {mejor_periodo['%_Aprobacion']:.2f}% ({mejor_periodo['Aprobados']}/{mejor_periodo['Inscriptos']})
   • Peor: {peor_periodo['Período']} - {peor_periodo['%_Aprobacion']:.2f}% ({peor_periodo['Aprobados']}/{peor_periodo['Inscriptos']})
""")

# Materias con mejor y peor aprobación
if len(materias_por_semestre) > 0:
    # Filtrar materias con al menos 10 estudiantes
    materias_filtradas = materias_por_semestre[materias_por_semestre['Inscriptos'] >= 10]
    
    if len(materias_filtradas) > 0:
        mejor_materia = materias_filtradas.loc[materias_filtradas['%_Aprobacion'].idxmax()]
        peor_materia = materias_filtradas.loc[materias_filtradas['%_Aprobacion'].idxmin()]
        
        print(f"""
📚 Mejor y Peor Materia (mínimo 10 estudiantes):
   • Mejor: {mejor_materia['Materia']} ({mejor_materia['Período']}) - {mejor_materia['%_Aprobacion']:.2f}% ({mejor_materia['Aprobados']}/{mejor_materia['Inscriptos']})
   • Peor: {peor_materia['Materia']} ({peor_materia['Período']}) - {peor_materia['%_Aprobacion']:.2f}% ({peor_materia['Aprobados']}/{peor_materia['Inscriptos']})
""")

print("="*80)
print("🎉 ¡ANÁLISIS COMPLETADO!")
print("="*80)
```
