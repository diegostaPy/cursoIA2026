## 1. El Pipeline de IA

**El pipeline de IA** es el flujo de trabajo completo que se sigue en un proyecto de inteligencia artificial:

```
[Definición del Problema] -> [Recolección de Datos] -> [EDA y Preprocesamiento]
                                                              |
                                                              v
[Despliegue y Monitoreo] <- [Evaluación y Ajuste] <- [Entrenamiento del Modelo]
```

**Etapas del pipeline:**

1. **Definición del Problema:** ¿Qué queremos predecir? ¿Cuál es el objetivo de negocio?
2. **Recolección de Datos:** Obtener los datos necesarios (bases de datos, APIs, archivos)
3. **EDA y Preprocesamiento:** Explorar, limpiar y preparar los datos para el modelo
4. **Entrenamiento del Modelo:** Seleccionar y entrenar el algoritmo de IA
5. **Evaluación y Ajuste:** Medir el rendimiento y optimizar hiperparámetros
6. **Despliegue y Monitoreo:** Poner el modelo en producción y supervisar su rendimiento

**En esta clase nos enfocaremos en las etapas de EDA, Preprocesamiento y Entrenamiento** usando Regresión Lineal.

## 3. Fundamentos de la Regresión Lineal

### 3.1 Modelo Lineal

**Modelo general:**

$$\hat{y} = w_1 x_1 + w_2 x_2 + \cdots + w_d x_d + b$$

**Forma vectorial:**

$$\hat{y} = \mathbf{w}^T \mathbf{x} + b$$

**Forma matricial (incluyendo bias en w):**

$$\mathbf{X} = \begin{bmatrix}
    1 & x_{11} & x_{12} & \cdots & x_{1d} \\
    1 & x_{21} & x_{22} & \cdots & x_{2d} \\
    \vdots & \vdots & \vdots & \ddots & \vdots \\
    1 & x_{N1} & x_{N2} & \cdots & x_{Nd}
\end{bmatrix}, \quad
\mathbf{w} = \begin{bmatrix} b \\ w_1 \\ w_2 \\ \vdots \\ w_d \end{bmatrix}, \quad
\mathbf{y} = \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_N \end{bmatrix}$$

$$\hat{\mathbf{y}} = \mathbf{X} \mathbf{w}$$

Donde $N$ es el número de muestras y $d$ es el número de características.

### 3.2 Función de Costo: Error Cuadrático Medio (MSE)

**Definición:**

$$J(\mathbf{w}) = \frac{1}{2N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2 = \frac{1}{2N} \sum_{i=1}^{N} (y_i - \mathbf{x}_i^T \mathbf{w})^2$$

**Forma matricial:**

$$J(\mathbf{w}) = \frac{1}{2N} (\mathbf{y} - \mathbf{X}\mathbf{w})^T (\mathbf{y} - \mathbf{X}\mathbf{w})$$

**Observación:** El factor $\frac{1}{2}$ se usa por conveniencia en la derivación (cancela el 2 al derivar).

**Objetivo:** Encontrar $\mathbf{w}$ que minimiza $J(\mathbf{w})$.

---

## 4. Derivación de la Ecuación Normal

### Paso 1: Expandir la función de costo

$$J(\mathbf{w}) = \frac{1}{2N} (\mathbf{y} - \mathbf{X}\mathbf{w})^T (\mathbf{y} - \mathbf{X}\mathbf{w})$$

**Expansión:**

$$J(\mathbf{w}) = \frac{1}{2N} \left[ \mathbf{y}^T \mathbf{y} - \mathbf{y}^T \mathbf{X}\mathbf{w} - (\mathbf{X}\mathbf{w})^T \mathbf{y} + (\mathbf{X}\mathbf{w})^T \mathbf{X}\mathbf{w} \right]$$

**Simplificando (términos escalares):**

$$J(\mathbf{w}) = \frac{1}{2N} \left[ \mathbf{y}^T \mathbf{y} - 2\mathbf{w}^T \mathbf{X}^T \mathbf{y} + \mathbf{w}^T \mathbf{X}^T \mathbf{X} \mathbf{w} \right]$$

- $\mathbf{y}^T \mathbf{y}$: constante (no depende de $\mathbf{w}$)
- $-2\mathbf{w}^T \mathbf{X}^T \mathbf{y}$: lineal en $\mathbf{w}$
- $\mathbf{w}^T \mathbf{X}^T \mathbf{X} \mathbf{w}$: cuadrático en $\mathbf{w}$

### Paso 2: Calcular el gradiente

**Reglas de derivación matricial:**
- $\nabla_{\mathbf{w}} (\mathbf{w}^T \mathbf{a}) = \mathbf{a}$
- $\nabla_{\mathbf{w}} (\mathbf{w}^T \mathbf{A} \mathbf{w}) = 2\mathbf{A}\mathbf{w}$ (si $\mathbf{A}$ es simétrica)

**Aplicando:**

$$\nabla_{\mathbf{w}} J(\mathbf{w}) = \frac{1}{2N} \left[ -2\mathbf{X}^T \mathbf{y} + 2\mathbf{X}^T \mathbf{X} \mathbf{w} \right] = \frac{1}{N} \left[ -\mathbf{X}^T \mathbf{y} + \mathbf{X}^T \mathbf{X} \mathbf{w} \right]$$

$$\nabla_{\mathbf{w}} J(\mathbf{w}) = \frac{1}{N} \left( \mathbf{X}^T \mathbf{X} \mathbf{w} - \mathbf{X}^T \mathbf{y} \right)$$

### Paso 3: Igualar el gradiente a cero y resolver

$$\nabla_{\mathbf{w}} J(\mathbf{w}) = 0$$

$$\frac{1}{N} \left( \mathbf{X}^T \mathbf{X} \mathbf{w} - \mathbf{X}^T \mathbf{y} \right) = 0$$

$$\mathbf{X}^T \mathbf{X} \mathbf{w} = \mathbf{X}^T \mathbf{y}$$

**Ecuación Normal:**

$$\boxed{\mathbf{X}^T \mathbf{X} \mathbf{w} = \mathbf{X}^T \mathbf{y}}$$

**Si $\mathbf{X}^T \mathbf{X}$ es invertible:**

$$\boxed{\hat{\mathbf{w}} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}}$$

### Paso 4: Verificación de la segunda derivada

$$\nabla_{\mathbf{w}}^2 J(\mathbf{w}) = \frac{1}{N} \mathbf{X}^T \mathbf{X}$$

$\mathbf{X}^T \mathbf{X}$ es semidefinida positiva $\Rightarrow$ mínimo global

## 5. Formulación Estadística: Máxima Verosimilitud (MLE)

### 5.1 Suposición fundamental

$$y = \mathbf{x}^T \mathbf{w} + \epsilon, \quad \epsilon \sim \mathcal{N}(0, \sigma^2)$$

### 5.2 Interpretación probabilística

$$p(y | \mathbf{x}, \mathbf{w}, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y - \mathbf{x}^T \mathbf{w})^2}{2\sigma^2}\right)$$

**Significado:**
- $y$ es una variable aleatoria con media $\mathbf{x}^T \mathbf{w}$
- El ruido $\epsilon$ es Gaussiano (Teorema Central del Límite)
- $\sigma^2$ es la varianza del ruido

### 5.3 Función de Verosimilitud

$$L(\mathbf{w}) = \prod_{i=1}^{N} p(y_i | \mathbf{x}_i, \mathbf{w}, \sigma^2)$$

**Log-verosimilitud:**

$$\ell(\mathbf{w}) = -\frac{N}{2}\log(2\pi\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^{N} (y_i - \mathbf{x}_i^T \mathbf{w})^2$$

### 5.4 Equivalencia MSE = MLE

Para $\sigma^2$ fijo, maximizar $\ell$ equivale a minimizar $\sum (y_i - \mathbf{x}_i^T \mathbf{w})^2$, que es exactamente el MSE.

$$\boxed{\hat{\mathbf{w}}_{\text{MLE}} = \hat{\mathbf{w}}_{\text{MSE}} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}}$$

**Conclusión clave:** Bajo ruido Gaussiano, MLE y MSE son equivalentes.

## 6. Gradiente Descendente (GD)

### 6.1 Algoritmo

El gradiente descendente es un método iterativo que actualiza los parámetros en la dirección del gradiente negativo de la función de costo:

$$\mathbf{w}^{(t+1)} = \mathbf{w}^{(t)} - \eta \nabla J(\mathbf{w}^{(t)})$$

Donde:
- $\eta$ es la tasa de aprendizaje
- $\nabla J(\mathbf{w}) = \frac{1}{N} \mathbf{X}^T (\mathbf{X}\mathbf{w} - \mathbf{y})$

### 6.2 Algoritmo en pseudocódigo

```
Inicializar w(0) (ej: 0)
Definir eta > 0 (tasa de aprendizaje)
t ← 0
Repetir:
    Calcular gradiente: ∇J(w(t)) = (1/N) * X^T (X w(t) - y)
    Actualizar: w(t+1) ← w(t) - eta * ∇J(w(t))
    t ← t + 1
Hasta convergencia
```

### 6.3 Interpretación geométrica

El gradiente apunta en la dirección de máximo crecimiento. Restamos el gradiente para movernos hacia el mínimo.

### 6.4 Gradiente Descendente Estocástico (SGD)

En SGD, se actualiza usando una sola muestra aleatoria en cada iteración:

$$\mathbf{w}^{(t+1)} = \mathbf{w}^{(t)} - \eta \nabla J_i(\mathbf{w}^{(t)})$$

Donde $\nabla J_i(\mathbf{w}) = \mathbf{x}_i^T (\mathbf{x}_i \mathbf{w} - y_i)$

**Ventajas de SGD:**
- Mucho más rápido para grandes datasets
- Permite entrenar con datos que no caben en memoria
- Puede escapar de mínimos locales

## 7. Métricas de Evaluación

### 7.1 Error Absoluto Medio (MAE)

$$MAE = \frac{1}{N} \sum_{i=1}^{N} |y_i - \hat{y}_i|$$

**Interpretación:** Error promedio en las mismas unidades que la variable objetivo.

### 7.2 Error Cuadrático Medio (MSE)

$$MSE = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$$

**Interpretación:** Penaliza más los errores grandes.

### 7.3 Raíz del Error Cuadrático Medio (RMSE)

$$RMSE = \sqrt{MSE} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2}$$

**Interpretación:** En las mismas unidades que la variable objetivo, más fácil de interpretar que MSE.

### 7.4 Coeficiente de Determinación (R²)

$$R^2 = 1 - \frac{\sum_{i=1}^{N} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{N} (y_i - \bar{y})^2}$$

**Interpretación:** Proporción de la varianza explicada por el modelo. Varía entre 0 y 1.

## 8. Selección de Características

**Objetivo:** Identificar qué variables predictoras son realmente útiles para el modelo.

**Método:** Entrenar modelos con diferentes conjuntos de características y comparar el RMSE.

**Ejemplo:** Eliminar `Newspaper` del modelo de ventas.

**Conclusión:** Si el RMSE disminuye al eliminar una variable, esa variable no es relevante para el modelo.

## 9. Análisis de Diagnóstico

### 9.1 Análisis de Residuos

**Definición de Residuos:**

$$e_i = y_i - \hat{y}_i$$

**Propiedades deseables:**
- Media cero: $\frac{1}{N} \sum e_i \approx 0$
- Distribución normal
- Homocedasticidad (varianza constante)

**Gráficos de diagnóstico:**
- Residuos vs valores predichos (buscar patrones)
- Histograma de residuos (buscar normalidad)

### 9.2 El Problema de las Escalas

**Número de condición:**

$$\kappa(\mathbf{X}^T \mathbf{X}) = \frac{\lambda_{max}}{\lambda_{min}}$$

- $\kappa \gg 1$: Matriz mal condicionada
- Causa inestabilidad numérica en la inversión
- Los coeficientes se vuelven no interpretables

### 9.3 Factor de Inflación de la Varianza (VIF)

$$VIF_j = \frac{1}{1 - R_j^2}$$

Donde $R_j^2$ es el coeficiente de determinación al regresar $X_j$ contra las demás variables.

**Interpretación:**
- $VIF > 10$: Multicolinealidad severa
- $VIF > 5$: Posible multicolinealidad
- $VIF < 5$: Aceptable

**Consecuencias de la multicolinealidad:**
- Coeficientes inestables
- Errores estándar inflados
- Dificultad para interpretar la importancia de las variables

## 10. Normalización de Datos

### 10.1 Técnicas de Normalización

**Normalización Min-Max:**

$$x' = \frac{x - x_{min}}{x_{max} - x_{min}} \in [0, 1]$$

**Estandarización Z-score:**

$$x' = \frac{x - \mu}{\sigma}$$

**¿Cuál usar?**
- Min-Max: Cuando se necesita un rango fijo (ej: [0,1])
- Z-score: Cuando los datos tienen outliers o se necesita comparar variables en diferentes escalas

### 10.2 Beneficios de la Normalización

1. **Estabilidad numérica:** Reduce el número de condición
2. **Convergencia más rápida:** En algoritmos iterativos como GD
3. **Interpretabilidad:** Los coeficientes se vuelven comparables
4. **Robustez:** Menos sensible a outliers

## 11. Resumen y Conclusiones

### 11.1 Conceptos Clave

1. **Ecuación Normal:** Solución directa pero costosa para grandes datasets
2. **Gradiente Descendente:** Iterativo, eficiente para grandes volúmenes
3. **MLE:** La regresión lineal minimiza MSE bajo ruido Gaussiano
4. **Número de Condición (κ):** Medida de estabilidad numérica
5. **VIF:** Detección de multicolinealidad
6. **Normalización:** Crucial para interpretabilidad y estabilidad

### 11.2 Regla de Oro

**Siempre preprocesar los datos antes de entrenar cualquier modelo.**

### 11.3 Comparación Sin vs Con Preprocesamiento

| Aspecto | Sin Preprocesamiento | Con Preprocesamiento |
|---------|---------------------|---------------------|
| Número de condición | $\kappa \gg 1$ | $\kappa \approx 1$ |
| Convergencia | Lenta | Rápida |
| Interpretabilidad | Pesos no comparables | Pesos comparables |
| Estabilidad | Inestable | Estable |
| Sensibilidad a outliers | Alta | Baja |
| Multicolinealidad | VIF alto | VIF reducido |
